```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# 1. Student ID last two digit (d)
d = 25
print(f"1. Student ID's last two digits (d): {d}\n")

# 2. Calculated Observed Frequencies
obs_c = 20 + d
obs_cpp = 25 + ((d % 10) * 25) + (d % 10)
obs_java = 30 + ((d % 7) * 30) + (d % 7)
obs_python = 35 + ((d % 5) * 35) + (d % 5)

observed = np.array([obs_c, obs_cpp, obs_java, obs_python])
languages = ['C', 'C++', 'Java', 'Python']

print("2. Calculated Observed Frequencies (O):")
for lang, val in zip(languages, observed):
    print(f"   - {lang}: {val}")
print(f"   Total Observed (N) = {sum(observed)}\n")

# 3 & 4. Hypotheses
print("3. Null Hypothesis (H0): Students have an equal preference for the four programming languages.")
print("4. Alternative Hypothesis (H1): Students do not have an equal preference for the four programming languages.\n")

# 5. Expected Frequency
total_n = sum(observed)
expected_val = total_n / len(languages)
expected = np.array([expected_val] * len(languages))
print(f"5. Expected Frequency (E) for each category: {expected_val}\n")

# 6, 7, 8. Statistics Calculation
chi2_stat, p_val = stats.chisquare(f_obs=observed, f_exp=expected)
dof = len(languages) - 1
critical_value = stats.chi2.ppf(1 - 0.05, dof)

print(f"6. Calculated Chi-Square statistic: {chi2_stat:.4f}")
print(f"7. Degrees of Freedom (df): {dof}")
print(f"8. Critical Value: {critical_value:.4f} | p-value: {p_val:.4e}\n")

# 9. Final Decision
print("9. Final Decision:")
if chi2_stat > critical_value:
    print("   Decision: Reject the Null Hypothesis (H0)\n")
else:
    print("   Decision: Fail to Reject the Null Hypothesis (H0)\n")

# 10. Visualization (Bar chart output)
x = np.arange(len(languages))
plt.figure(figsize=(7, 4.5))
plt.bar(x - 0.2, observed, width=0.4, label='Observed (O)', color='#1f77b4', edgecolor='black')
plt.bar(x + 0.2, expected, width=0.4, label='Expected (E)', color='#ff7f0e', alpha=0.8, edgecolor='black')
plt.xticks(x, languages)
plt.xlabel('Programming Languages')
plt.ylabel('Frequencies')
plt.title(f'Chi-Square Test (d = {d})')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

1. My Student ID's last two digits:
My student ID's last two digits are d = 25.
2. The calculated Observed Frequencies (O):
By putting my ID digits into the given formulas, I got the following counts:

* C = 20 + 25 = 45
* C++ = 25 + (5 * 25) + 5 = 155
* Java = 30 + (4 * 30) + 4 = 154
* Python = 35 + (0 * 35) + 0 = 35
* Total Sample Size (N) = 45 + 155 + 154 + 35 = 389

3. The Null Hypothesis (H0):
H0: Students have an equal preference for the four programming languages.
4. The Alternative Hypothesis (H1):
H1: Students do not have an equal preference for the four programming languages.
5. The Expected Frequency (E) for each category:
Assuming that preferences are equal for all 4 categories, the expected frequency is:
E = Total (N) / 4 = 389 / 4 = 97.25
6. The calculated Chi-Square statistic:
Based on the Python script output, the calculated value is:
Calculated Chi-Square = 135.3290
7. The Degrees of Freedom:
df = number of categories - 1 = 4 - 1 = 3
8. The Critical Value or p-value:
At the 5% significance level (alpha = 0.05) with df = 3:

* Critical Value = 7.8147
* p-value = 3.95e-29 (which is extremely small and close to 0)

9. My final decision:
Reject the Null Hypothesis (H0).
Reason: The calculated Chi-Square value (135.3290) is much larger than the critical value (7.8147), and the p-value is way lower than 0.05.
10. A brief conclusion in the context of the scenario:
In conclusion, there is strong evidence to show that the university students do not have an equal preference for these four programming languages. C++ and Java are clearly more popular among them.
