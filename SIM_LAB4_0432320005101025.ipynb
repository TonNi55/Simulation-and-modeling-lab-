```python
import random
import numpy as np
import math

# Student information
student_id = "0432320005101025"
full_name = "Tamanna Afrin"

# Last two digits of Student ID
R = int(student_id[-2:])

# Range and radius
radius = R / 10

# Count letters in full name, ignoring space
number_of_letters = sum(char.isalpha() for char in full_name)

# Minimum number of points
minimum_points = 200 * number_of_letters

print("Student ID:", student_id)
print("Full Name:", full_name)
print("R:", R)
print("Range: [0,", radius, "]")
print("Radius:", radius)
print("Number of letters:", number_of_letters)
print("Minimum points:", minimum_points)
```

```python
def simulate(num_points):
    inside = 0

    # Half points using random.uniform()
    for i in range(num_points // 2):
        x = random.uniform(0, radius)
        y = random.uniform(0, radius)

        if x*x + y*y <= radius*radius:
            inside += 1

    # Other half using numpy.random.rand()
    for i in range(num_points - num_points // 2):
        x = np.random.rand() * radius
        y = np.random.rand() * radius

        if x*x + y*y <= radius*radius:
            inside += 1

    return inside
```

```python
# Run the simulation with increasing number of points

print("Range Used | Total Points | Points Inside | Estimated Pi | Error")
print("-" * 70)

for multiplier in [1, 2, 3, 4, 5]:

    num_points = minimum_points * multiplier

    inside = simulate(num_points)

    pi = 4 * inside / num_points

    error = abs(pi - math.pi)

    print(
        "[0,", radius, "] |",
        num_points, "|",
        inside, "|",
        round(pi, 6), "|",
        round(error, 6)
    )
```
