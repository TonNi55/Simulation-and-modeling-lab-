```python
import numpy as np

# Create a 3x2 matrix
matrix = np.array([[2,4],
                   [6,8],[10,15]])

# Apply trigonometric functions
print(np.sin(matrix))
print(np.cos(matrix))
print(np.tan(matrix))

# Apply inverse trigonometric functions
print(np.arcsin(np.clip(matrix/12,-1,1)))
print(np.arccos(np.clip(matrix/12,-1,1)))
print(np.arctan(matrix))

# Apply exponential and logarithmic functions
print(np.exp(matrix))
print(np.log(matrix))

# Find absolute value and square root
print(np.abs(matrix))
print(np.sqrt(matrix))

# Find remainder when divided by 2
print(np.remainder(matrix,2))

# Apply rounding functions
print(np.round(matrix))
print(np.floor(matrix))
print(np.ceil(matrix))
```

```python
import numpy as np

# Create a 2x3 matrix with float values
vec = np.array([[2.1,4.7,6.3],
                [8.5,1.2,3.9]])

# Find the maximum value
print(np.max(vec))

# Find the index of the maximum value
print(np.argmax(vec))

# Find the minimum value
print(np.min(vec))

# Find the index of the minimum value
print(np.argmin(vec))

# Find the vector length (number of rows)
print(len(vec))

# Sort each row in descending order
print(np.sort(vec)[:,::-1])

# Sort the entire matrix in descending order
print(np.sort(vec.flatten())[::-1])

# Calculate the sum of all elements
print(np.sum(vec))

# Calculate the product of all elements
print(np.prod(vec))

# Calculate the median
print(np.median(vec))

# Calculate the mean
print(np.mean(vec))

# Calculate the standard deviation
print(np.std(vec))
```
