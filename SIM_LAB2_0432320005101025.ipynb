```python
import numpy as np

# Generate 1,000 wait times for Regular Days
regular_days = np.random.uniform(3, 15, 1000)

# Generate 1,000 wait times for Busy Days
busy_days = np.random.normal(3, 15, 1000)

# Replace negative values with 0
busy_days[busy_days < 0] = 0

print("Regular Days Data Generated:", len(regular_days))
print("Busy Days Data Generated:", len(busy_days))
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Generate 1000 random values from a normal distribution
SNormalValues = np.random.randn(1000)#mean=9, standard deviation=2

# Create a histogram
plt.hist(SNormalValues, bins=30, edgecolor='black')
plt.title('Histogram of Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python

# With Seed
np.random.seed(4)  # Setting seed to 10
seeded_matrix = np.random.randint(1, 51, (3, 3))
print("Seeded Random Matrix:\n", seeded_matrix)

# Without Seed
np.random.seed(None)  # No seed, generates different values each time
non_seeded_matrix = np.random.randint(1, 51, (3, 3))
print("\nNon-Seeded Random Matrix:\n", non_seeded_matrix)
```
