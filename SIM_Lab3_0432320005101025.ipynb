```python
import numpy as np

# Generate 3,000 wait times for Regular Days
regular_days = np.random.uniform(2, 12, 3000)

# Generate 3,000 wait times for Busy Days
busy_days = np.random.normal(8, 2, 3000)

# Replace negative values with 0
-
busy_days[busy_days < 0] = 0

print("Regular Days Data Generated:", len(regular_days))
print("Busy Days Data Generated:", len(busy_days))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Regular Days
plt.subplot(1, 2, 1)
plt.hist(regular_days, bins=30, edgecolor='black')
plt.title("Regular Days - Uniform Distribution")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

# Busy Days
plt.subplot(1, 2, 2)
plt.hist(busy_days, bins=30, edgecolor='black')
plt.title("Busy Days - Normal Distribution")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Regular Days Statistics
print("===== REGULAR DAYS =====")
print("Mean:", np.mean(regular_days))
print("Median:", np.median(regular_days))
print("Standard Deviation:", np.std(regular_days))
print("Maximum Wait Time:", np.max(regular_days))

# Busy Days Statistics
print("\n===== BUSY DAYS =====")
print("Mean:", np.mean(busy_days))
print("Median:", np.median(busy_days))
print("Standard Deviation:", np.std(busy_days))
print("Maximum Wait Time:", np.max(busy_days))
```
